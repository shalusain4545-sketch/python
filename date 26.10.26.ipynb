```python
class Student:
    def __init__(self,name,age):
        self.name=name  #instance variable
        self.age=age  # Instance variable

# creating objects
s1 = Student("Rahul",20)
s2 = Student("Priya",22)
 
print(s1.name)
print(s1.age)

print(s2.name)
print(s2.age)
```

```python
class Student:
    
   def __init__(self,name,marks):
    self.name = name
    self.marks = marks

   def display(self): # instance method
    print("Name:",self.name)
    print("Marks:",self.marks)


s1= Student("Rahul",85)
s2= Student("Priya",92)


s1.display()
s2.display()
```

```python
class Student:
    school_name ="ABC School"

    def __init__(self,name):
        self.name = name

@classmethod
def change_school(new_school):
     Student.school_name=new_school    


s1 = Student("Rahul")
s2 = Student("Priya")


print(s1.school_name)
print(s2.school_name)

Student.change_school("XYZ School")

print(s1.school_name)
print(s2.school_name)
```

```python
class Student:

    @staticmethod
    def is_passed(marks):
        if marks>= 40:
            return True
        else: 
            return False


print(Student.is_passed(75))
print(Student.is_passed(30))
```
