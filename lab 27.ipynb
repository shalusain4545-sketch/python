```python
# Number of rows for the triangle 
```

```python
rows = 5 
```

```python
# outer loop for the number of rows
```

```python
for i in range(1,rows + 1):
    # Inner loop for printing asterisks in each row
    for j in range(i):
        print("*", end=" ") # print an asterisk followed by a space
    print() # Move to the next line after finishing the inner loop     
```

```python
# for loop with break 
```

```python
# Using break to exit the loop when a condition is met
for i in range (10):
    if i == 5:
        break
    print(i)    
```

```python
# for loop with continue
```

```python
# Skipping even numbers
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)    
```

```python
# for loop with else 
```

```python
# Using else with a for loop
for i in range(5):
    print(i)
else:
    print("Loop completed successfuly.")
```
