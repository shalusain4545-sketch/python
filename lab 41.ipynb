```python
def add(x,y):
    """Returns the sum of two numbers."""
    print( x + y) 
def subtract(x,y):
    """Returns the difference between two numbers."""
    print (x-y)
def multiply(x,y):
    """Returns the product of two numbers."""
    print (x *y)
    
def divide(x,y):
    """Returns the quotient of two numbers.Raises ZeroDivisionError if the divisor is zero."""
    if y == 0:
        raise ValueError("Cannot divide by Zero.")
    print (x / y)

add(5,6)
subtract(5,6)
multiply(5,6)
divide(10,2)
```
