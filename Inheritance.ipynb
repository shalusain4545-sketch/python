```python
class parent: # singal level
    def display_detail(self):
        print("Hello world")
class child(parent):
    def add(self,a,b):
        print(a+b)
    def display_detail(self):
        print("greet")
c=child()
c.add(5,10)
c.display_detail()

p=parent()
p.display_detail()         
            
```

```python
class gf: # multi level
    def add(self,a,b):
        print(a+b)

class parent(gf):
    def sub(self,c,d):
        print(c-d)
class child(parent):
    def mul(self,e,f):
        print(e*f)
c=child()
c.add(5,10)
c.sub(10,5)
c.mul(5,50)
```

```python
class parent1: # multiple 
    def add(self,a,b):
        print(a+b)

class parent2:
    def sub(self,c,d):
        print(c-d)
class child(parent1,parent2):
    def mul(self,e,f):
        print(e*f)
c=child()
c.add(5,10)
c.sub(10,5)
c.mul(5,50)
```

```python
class parent1: #Herarchial
    def add(self,a,b):
        print(a+b)
        self.a=a
        self.b=b

class child1(parent):
    def sub(self,c,d):
        print(c-d)
class child2(parent):
    def mul(self,e,f):
        print(e*f)
c1=child()
c1.add(5,10)
c1.sub(10,5)
c1.mul(5,50)
c2=child2()
c2.add(10,50)
c2.sub(50,10)
c2.mul(50,50000)
```
