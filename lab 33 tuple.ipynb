```python
# Task 1: implementation of creating a tuple 
# Creating a tuple
my_tuple = (10,20,30,40,50,60)
print("Tuple:",my_tuple)
```

```python
#Task 2: implementation of Accessing Elements in a Tuple
print("First element:", my_tuple[0])
print("Last element:", my_tuple[-1])
```

```python
# Task 3:implementation of slicing a tuple
print("Slice [1:4]:", my_tuple[1:4])
print("Slice from beginning to index 3:", my_tuple[:3])
print("Slice from index 2 to end:",my_tuple[2:])
```

```python
# Task 4:implementation of concatenating two tuples
extra_tuple =(70,80,90)
concatenated = my_tuple + extra_tuple
print("Concatenated Tuple:",concatenated)
```

```python
#Task 5: implementation of iterating Through a tuple
for item in my_tuple:
    print("Item:",item)
```

```python
# Task 6:implementation of membership test in a tuple
print("Is 30 in tuple?",30 in my_tuple)
print("Is 100 not in tuple?",100 not in my_tuple)
```

```python
# Task 7: implementation of Nested Tuples and Accessing Their Elements
nested_tuple=(my_tuple,(70,80,90))
print("Nested Tuple:",nested_tuple)
print("Accessing nested element (my_tuple[2]):",nested_tuple[0][2])
print("Accessing nested element(90):",nested_tuple[1][2])
```

```python
# Task 8:implementation of Applying Built-in Functions on Tuples 
print("length of tuple:",len(my_tuple))
print("Maximum value:",max(my_tuple))
print("Minimum value:",min(my_tuple))
print("Sum of elements:",sum(my_tuple))
```

```python
# Task 9: implementation of updating a Tuple(using list conversion)
temp_list= list(my_tuple)
temp_list[2]=99 # update 3rd element
updated_tuple = tuple(temp_list)
print("Updated Tuple:",updated_tuple)
```
