```python
import requests
from bs4 import BeautifulSoup

def scrape_example():
    url= 'https://edunetfoundation.org/'
    response = requests.get(url)
    if response.status_code == 200:
       soup = BeautifulSoup(response.text,'html.parser')
       title = soup.title.get_text()
       print(f'Title of the page:{title}')
    else:
        print(f"Failed to retrieve the page.Status code:{response.status_code}")
if __name__ == "__main__":
    scrape_example()
```

```python
# Import the requests library
import requests
# define the url of the website to scrape
URL = "https://edunetfoundation.org/"
# send a get request to the specified url and store the response in 'resp'
resp = requests.get(URL)
# print the HTTP status code of the response to check if request was successful
print("Statys Code:",resp.status_code)
# print the HTML content of the response 
print("\nResponse Content:")
print(resp.text)
```
