```python
print(f"The result of {15}+{4} is:{15+4}")
print(f"The result of {15}-{4} is:{15-4}")
print(f"The result of {15}*{4} is:{15*4}")
print(f"The result of {15}/{4} is:{15/4}")
print(f"The result of {15}**{4} is:{15**4}")
print(f"The result of {15}//{4} is:{15//4}")
```

```python
# Inserting space between outputs using commas 
print("The sum of 5 and 3 is",5 + 3)
print("The product of 4 and 7 is",4 * 7)
```

```python
# Custom separator between outputs
print("5", "+","3","=",5 + 3,sep=" ")
print("4", "*","7","=", 4 * 7,sep=" * ")
```

```python
# Inserting text at the end of a print statement using end parameter
print("5 + 3 =",end=" ")
print(5+3)
print("4 * 7 =",end=" ")
print(4 * 7)
```
