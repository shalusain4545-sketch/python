```python
# task 1. local scope

def local_scope_example():
    x = 10 # local variable
    print("Local Variable x:",x)

local_scope_example()
# print(x)
# This will raise a NameError because x is not accessiable outside the function
```

```python
# task 2.Enclosing scope (Nested Functions)

def outer_function():
    x=10 # Enclosing variable

    def inner_function():
        print("Enclosing variable x from inner function:",x)

    inner_function()

outer_function()     
```

```python
# task 3. Global scope
x = 10 # Global variable

def global_scope_example():
    print("Global variable x:",x)

global_scope_example()
print("Global variable x outside function:",x)
```

```python
# task 4. Modifying Global variables
x = 10 # Global variable

def Modify_global ():
    global x
    x = 20 # modifying Global variable

Modify_global()
print("Global variable x after modification:",x)
```
