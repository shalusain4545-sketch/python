```python
my_set={1,2,3,4,5} # Adding elements to a set 
```

```python
my_set.add(6)
```

```python
print(my_set)
```

```python
my_set.update({7,8,9})# adding multiple elements to a set 
print(my_set)
```

```python
# Removing elements from  set 
my_set.remove(3)
print(my_set)
```

```python
# Siscarding elements from s set ( no error if element not present)
my_set.discard(10)
print(my_set)
```

```python
# lterating over a set 
for item in my_set:
    print(item)
```

```python
# length of a set 
print(len(my_set))
```

```python
# using curly braces
fruits = {"apple","banana","cherry"}
print(fruits)
```

```python
# Adding elements to a set
# add a single element
fruits.add("orange")
print(fruits)
```

```python
# add multiple elements
fruits.update(["mango","grapes"])
print(fruits)
```

```python
# creating a dictionary
country_capitals= {"Germany:Berlin","Canada:Ottawa","England:London"}
# printing the dictionay 
print(country_capitals)
```

```python
# creating a dictionary
my_dict={"name":"john","age":30,"city":"New York"}
# Accessing values in a dictionary using keys
print(my_dict["name"])
print(my_dict["age"])
print(my_dict["city"])
```

```python
# Modifying values in a dictionary
my_dict ={
    "name":"john","age":30,
    "city":"New York"}
my_dict["age"] =35
```

```python
print(my_dict)
```

```python
# Adding new key-value pairs to a dictionary
my_dict["gender"]="Male"
print(my_dict)
```

```python
# Removing key-value pairs from a dictionary
del my_dict["city"]
print(my_dict)
```

```python
# length of a dictionary
print(len(my_dict))
```

```python
# checking for existence of keys in a dictionary
print("name"in my_dict)
print("city"in my_dict)
```

```python
print(my_dict.keys())
print(my_dict.values())
```
