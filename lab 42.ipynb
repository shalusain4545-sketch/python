```python
# Task 1: implementation of math module
# math Module
import math

# square root
print("Square root of 25:",math.sqrt(25))

# Factorial
print("Factorial of 5:",math.factorial(5))

# Value of pi
print("Value of pi:",math.pi)

# cosine of 45 degrees
print("Cosine of 45 degrees:",math.cos(math.radians(45)))
```

```python
# Task 2:implementation of datetime module
# datetime module
import datetime
# current date and time 
now = datetime.datetime.now()
print("Current date and time:",now)

# specific date
d=datetime.date(2024,8,27)
print("Specific date:",d)

# difference between two dates
delta = now - datetime.datetime(2024,1,1)
print("Days since January 1,2024:",delta.days)

# formatting dates
formatted_date = now.strftime("%Y-%m-%d %H:%M:%S")
print("Formatted date and time:",formatted_date)
```

```python
# task 3 implementation of os module
# os module
import os

# current working directory
print("Current working directory:",os.getcwd())

# list files in the current directory 
print("Files in current directory:",os.listdir('.'))

# create a new directory
os.makedirs('new_directory',exist_ok=True)
print("Directory 'new_directory'created.")

# Remove the directory
os.rmdir('new_directory')
print("Directory 'new_directory'removed.")

```

```python
# task implementation of os sys module
# sys Module
import sys
# print python version
print("python version:",sys.version)

# print system platform
print("platform:",sys.platform)

# print command line arguments
print("Command line arguments:",sys.argv)
```

```python
# task 5 implementation of os random module
# random module
import random

# Random integer between 1 and 10
print("Random integer between 1 and 10:",random.randint(1,10))
# Random flooting point number between 0 and 1
print("Random float between 0 and 1:",random.random())

# Random choise from a list 
items=['apple','banana','cherry']
print("Random choice from list:",random.choice(items))
```
