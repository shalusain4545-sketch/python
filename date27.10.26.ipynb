```python
class Student:
    def __init__(self,name,marks):
        self.name = name # public attribute
        self.marks = marks # public attrubute

    def display(self):    # public method
        print("Name",self.name)
        print("Marks:",self.marks)

# creating an object
s1 = Student("Shalu",90)
# Accessing publicc attrubutes
print(s1.name)
print(s1.marks)

# Calling a public method
s1.display()
        
```

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self._salary = salary

    def _display_salary(self):
        print("Salary:",self._salary)


class Manager(Employee):
    def show_details(self):
        print("Name:",self.name)


        print("Salary:",self._salary)
        
        self._display_salary()

m1=Manager("Shalu",50000)

m1.show_details()
    
```

```python
class BankAccount:
    def __init__(self,name,balance,pin):
        self.name = name # public attribute
        self.__balance = balance # private attribute
        self.__pin = pin

    def __verify_pin(self,entered_pin): # private method
        return self.__pin == entered_pin

    def show_balance(self,entered_pin):
        if self.__verify_pin(entered_pin):
            print("Balance:",self._balance)

        else:  
            print("Incorrect PIN")

account = BankAccount("Shalu",10000,1234)

print(account.name)

account.show_balance(12345)
print(account.__balance)

            
            
```

```python
class person:
    def __init__(self,name):
       self._name = name # private attribute

# Getter method to access private attribute
    def get_name(self):
         return self._name

# Setter method to modify private attribute 
    def set_name(self,name):
        self._name = name

# Creating an objet of the person class 
person = person("Chandni")

# Accessing private data using the getter method
print(person.get_name())

# Modifying private data using the setter method

person.set_name("Shalu")

# Accessing modified data
print(person.get_name())
    
```
