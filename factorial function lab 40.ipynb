```python
# Factorial calculation
def factorial(n):
    # Base case
    if n == 0:
        return 1
    # Recursive case
    else:
        return n* factorial(n-1)

# lest the factorial function
n= int(input("Enter N:"))
print("Factorial of 5:", factorial(5))
        
```

```python
# Fibonacci Sequence
def fibonacci(n):
    # Base cases
    if n <= 0:
        return 0
    elif n==1:
        return 1
# Recursive case
    else:
       return fibonacci(n-1)+ fibonacci(n-2)
# test the fibonacci function
n = int(input("Enter N:"))
print("Fibonacci number at position",n,":",fibonacci(n))
```
