```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse


def scrape_title(url):
    try:
        # Validate the URL
        parsed_url = urlparse(url)

        if parsed_url.scheme not in ("http", "https") or not parsed_url.netloc:
            print("Error: Please enter a valid URL.")
            return

        # Send request to the website
        headers = {
            "User-Agent": "Mozilla/5.0"
        }

        response = requests.get(
            url,
            headers=headers,
            timeout=15
        )

        response.raise_for_status()

        # Parse the HTML content
        soup = BeautifulSoup(response.text, "html.parser")

        # Extract the page title
        title_tag = soup.find("title")

        if title_tag and title_tag.get_text(strip=True):
            title = title_tag.get_text(strip=True)
            print("\nWeb Page Title:", title)
        else:
            print("\nNo title found on this web page.")

    except requests.exceptions.Timeout:
        print("Error: The website took too long to respond.")

    except requests.exceptions.RequestException as error:
        print("Error fetching the webpage:", error)


def main():
    print("=" * 40)
    print("       WEB PAGE TITLE SCRAPER")
    print("=" * 40)

    url = input("Enter website URL: ").strip()

    if url:
        scrape_title(url)
    else:
        print("Error: URL cannot be empty.")


if __name__ == "__main__":
    main()
```

```python
import requests
from bs4 import BeautifulSoup

url = input("Enter website URL: ")

res = requests.get(url)
soup = BeautifulSoup(res.text, "html.parser")

print("Web Page Title:", soup.title.string)
```
