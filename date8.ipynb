```python
# prompt the user to input a temprature in the format (e.g. ,45F,102C, etc.) lab 25
tem = input("input the temperature you like to convert (e.g. ,45F,102C,etc.):")

# extract the bnumerical part of the temperature and convert it to an integer
degree = int(tem[:-1])

# extract the convention part of the temperature input (either 'C' or 'F')
i_convention = tem[-1]

#check if the input convention is in uppercase 'C'(Celsius)
if i_convention.upper()=="C":
    # convert the celsius temperature to Fahrenheit
    result = int(round((9* degree) / 5 + 32))
    o_convention = "Fahrenheit" # set the output convention as Fahrenheit
# check (f the input convention is in uppercase 'F' (Fahrenheit)
elif i_convention.upper() == "F":
    # convert the Fahrenhiest tempreature to Celsius 
    result = int(round((degree-32)*5/9))
    o_convention = "Celsius"   #set the output convention as Celsius
else:
    #if the input convenation is neithr'C'nor 'F',print an error message and exit the program
    print("input proper convention.")
    quit()
#display the converted tempreaturei in the specified output convention
print("the tempreature in",o_convention,"is",result,"degree.")
```
