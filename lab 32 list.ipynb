```python
#creating a list
my_list=[1,2,3,4,5]
```

```python
# Accessing elements of a list
print(my_list[0])
print(my_list[-1])
print(my_list[2:4])
```

```python
# Modifying elements of a list
my_list[0]=10
print(my_list)
```

```python
# Adding elements to a list
my_list.append(6)
print(my_list)
```

```python
# lterating over a list
for item in my_list:
    print(item)
```

```python
# length of a list
print(len(my_list))
```

```python
# Creating a tuple
my_tuple =(1,2,3,4,5)
```

```python
# Accessing elements of a tuple 
# Since tuples are immutable,modifying elements not allowed
# This will raise an error:TypeError:'tuple object do not support item assignment
# my_tuple[0]=10
print(my_tuple[0])
print(my_tuple[-1])
print(my_tuple[2:4])      
```
